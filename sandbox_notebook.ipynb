```python
# Count candidates in the file
with open(candidate_file) as f:
    n = sum(1 for _ in f)
print(f"Candidates in file: {n}")

# Adjust output count if less than 100 candidates
output_count = min(n, 100)
print(f"Will output top {output_count} candidates")
```
